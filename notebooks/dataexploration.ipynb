```python
main_datasets = ["colleges", "universities", "standalone", "research"]

code_sets = {
    name: set(datasets[name]["Aishe Code" if name != "research" else "AISHE Code"]
              .dropna().astype(str).str.strip())
    for name in main_datasets
}

vidya_codes = set(
    datasets["vidya_lakshmi"]["AISHE Code"]
    .dropna().astype(str).str.strip()
)

for name, codes in code_sets.items():
    overlap = codes.intersection(vidya_codes)

    print(
        f"{name:<15} | "
        f"Overlapping Vidya Lakshmi records: {len(overlap)}"
    )
```

```python
all_main_codes = set.union(*code_sets.values())

vidya_df = datasets["vidya_lakshmi"]

unmatched_vidya = vidya_df[
    ~vidya_df["AISHE Code"]
    .astype("string")
    .str.strip()
    .isin(all_main_codes)
]

print("Total Vidya Lakshmi records:", len(vidya_df))
print("Matched with main directories:", len(vidya_df) - len(unmatched_vidya))
print("Unmatched records:", len(unmatched_vidya))

display(unmatched_vidya)
```

```python
for name, df in datasets.items():
    if "Year Of Establishment" not in df.columns:
        continue

    years = pd.to_numeric(
        df["Year Of Establishment"],
        errors="coerce"
    )

    invalid = years[
        (years < 1800) | (years > 2026)
    ]

    print(f"\n{name.upper()}")
    print("Missing or non-numeric:", years.isna().sum())
    print("Invalid years:", len(invalid))

    if not invalid.empty:
        display(df.loc[invalid.index, ["Name" if "Name" in df.columns else "Institute Name", "Year Of Establishment"]].head(10))
```

```python
for name, df in datasets.items():
    if "Year Of Establishment" not in df.columns:
        continue

    years = pd.to_numeric(
        df["Year Of Establishment"],
        errors="coerce"
    )

    invalid_mask = years.notna() & (
        (years < 1800) | (years > 2026)
    )

    if invalid_mask.any():
        print(f"\n{name.upper()} — Invalid years")
        display(
            df.loc[
                invalid_mask,
                ["Name" if "Name" in df.columns else "Institute Name",
                 "Year Of Establishment"]
            ]
        )
```

```python
for name, df in datasets.items():
    if "Website" not in df.columns:
        continue

    websites = df["Website"].astype("string").str.strip()

    missing = (
        websites.isna() |
        websites.eq("") |
        websites.eq("-")
    )

    print(f"\n{name.upper()}")
    print("Total records:", len(df))
    print("Missing websites:", missing.sum())
    print("Available websites:", (~missing).sum())
    print("Website coverage:", round((~missing).mean() * 100, 2), "%")
```

```python
for name, df in datasets.items():
    if "Website" not in df.columns:
        continue

    websites = df["Website"].dropna().astype(str).str.strip()
    websites = websites[~websites.isin(["", "-"])]

    missing_protocol = websites[
        ~websites.str.match(r"^https?://", case=False)
    ]

    print(f"\n{name.upper()}")
    print("Total available websites:", len(websites))
    print("Missing URL protocol:", len(missing_protocol))

    display(missing_protocol.head(5))
```


## Data Exploration Summary

### 1. Datasets
Explored five AISHE datasets:

| Dataset | Records | Columns |
|---|---:|---:|
| Colleges | 54,571 | 12 |
| Universities | 1,438 | 7 |
| Standalone | 17,092 | 8 |
| R&D Institutes | 291 | 6 |
| Vidya Lakshmi | 1,452 | 9 |

**Total records:** 74,844

### 2. Key Observations
- Identified institution details such as AISHE code, name, state, district, website and establishment year.
- College records include management, college type and affiliated university information.
- Standalone records include institution categories.
- R&D records include administrative ministry details.
- Vidya Lakshmi contains institutions overlapping with the other directories.

### 3. Data Quality Findings
- Institution names are not unique; AISHE codes will be used as identifiers.
- One repeated AISHE code was identified in the R&D dataset.
- Missing values and `-` placeholders exist in some columns.
- Some establishment years contain invalid values, including zero and 2027.
- Website URLs have inconsistent formats.
- Some records have missing district or website information.

### 4. Decisions
- Retain all five datasets.
- Preserve AISHE codes as primary identifiers.
- Keep Vidya Lakshmi as a supplementary dataset.
- Do not delete records solely because institution names are duplicated.
- Handle missing values and normalize fields during preprocessing.
- Preserve the original datasets.
